```python
# === Gráfica del vuelo del cohete de agua (CSV del altímetro ESP32) ===
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# 1. Subir el archivo CSV descargado de la ESP32
subido = files.upload()                      # Se abre el botón "Elegir archivos"
nombre = list(subido.keys())[0]

# 2. Leer los datos (columnas: tiempo_s, altura_m)
df = pd.read_csv(nombre)

# 3. Localizar el apogeo
i_max = df["altura_m"].idxmax()
t_apo, h_apo = df.loc[i_max, "tiempo_s"], df.loc[i_max, "altura_m"]

# 4. Graficar
plt.figure(figsize=(10, 5))
plt.plot(df["tiempo_s"], df["altura_m"], color="#2f6fb3", lw=2, label="Altura")
plt.scatter(t_apo, h_apo, color="#1f9d55", zorder=5)
plt.annotate(f"Apogeo: {h_apo:.1f} m @ {t_apo:.2f} s",
             (t_apo, h_apo), textcoords="offset points", xytext=(10, 8))
plt.title("Vuelo del cohete de agua — Altura vs. tiempo")
plt.xlabel("Tiempo (s)")
plt.ylabel("Altura (m)")
plt.grid(alpha=0.3)
plt.legend()
plt.show()

print(f"Apogeo: {h_apo:.2f} m  |  Duración registrada: {df['tiempo_s'].iloc[-1]:.2f} s")
```
